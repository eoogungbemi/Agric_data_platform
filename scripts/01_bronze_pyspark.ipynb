```python
from pyspark.sql import SparkSession

# python -c "from pyspark.sql import SparkSession; spark = SparkSession.builder.appName('test').getOrCreate(); print(spark.range(5).collect()); spark.stop()"

spark = (SparkSession.builder
         .appName("ARE_Raw_Ingest")
         .getOrCreate())

raw_input_path = "/Users/dev/PycharmProjects/jup/mnt/agri/landing"
raw_output_path = "/Users/dev/PycharmProjects/jup/mnt/agri/bronze/parquet"

farms_df = (spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(f"{raw_input_path}/farms.csv"))

sensors_df = (spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(f"{raw_input_path}/sensor_readings.csv"))

# Minimal checks / schema print
farms_df.printSchema()
sensors_df.printSchema()

# Persist as raw parquet
(farms_df
 .write
 .mode("overwrite")
 .parquet(f"{raw_output_path}/farms"))

(sensors_df
 .write
 .mode("overwrite")
 .parquet(f"{raw_output_path}/sensor_readings"))

spark.stop()
```
